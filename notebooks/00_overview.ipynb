# Uncertainty-Aware Adaptive QUBO for Real-World EV Charging

**Notebook 00 — Overview and reading guide.**

## Research question

Does an uncertainty-aware QUBO formulation with a calibration-driven adaptive deadline penalty (ADOPT) improve the operational reliability of EV charging schedules under behavioral uncertainty, when evaluated through exact optimization and QAOA in a small-instance regime, on a temporally held-out test set?

## Hypothesis

A scenario-averaged robust QUBO (F1) is expected to dominate a deterministic QUBO (F0) on held-out feasibility. The adaptive variant F2 (= F1 with ρ_d scaled by γ, computed from the calibration joint) is expected to dominate F1. The oracle F3 (which sees the realized uncertainty at evaluation time) provides an upper bound on what any deployable method could achieve; it is **analysis only**, not deployable.

## Limitations

The 11-qubit instance is small. The result does not predict behavior on larger instances. The Caltech site is one of several ACN-Data sites; the result may not generalize.


## Quantum-advantage disclaimer

This work does **not** claim quantum advantage. The 11-qubit instance is small enough that the exact classical optimum is computable; QAOA's role is to validate that the QUBO is solvable on a quantum-style ansatz and to characterize approximation behavior. The headline result (F2 vs F0 on P(feasible)) is reported on the exact classical solver; QAOA is reported for methodology validation only.


## No post-hoc tuning

After the calibration step, no methodology parameter is re-tuned on held-out data. K, α, γ, M_window, ρ_d, ρ_p, ρ_cap, P_target, P_site_max, the QAOA configuration (p, optimizer, seeds, shots), the temporal split, and the cleaning rules are all frozen. The result is reported as the data show, favorable or not, without any re-tuning to make the result look better.
